# Subsetting

We have already done some subsetting when we selected particular rows and columns. However, we often want to do more than select data by position. We may want to create a subset containing only observations that meet particular criteria.

Suppose we want a new data frame containing Gentoo penguins weighing less than 4,300 grams:

```r
lowGentoo <- penguins[penguins$species == "Gentoo" & penguins$body_mass < 4300,]

print(lowGentoo)
```

```text
    species island bill_len bill_dep flipper_len body_mass    sex year
167  Gentoo Biscoe     45.8     14.6         210      4200 female 2007
169  Gentoo Biscoe     42.0     13.5         210      4150 female 2007
179  Gentoo Biscoe     44.5     14.3         216      4100   <NA> 2007
193  Gentoo Biscoe     42.7     13.7         208      3950 female 2008
199  Gentoo Biscoe     45.5     13.9         210      4200 female 2008
213  Gentoo Biscoe     45.3     13.8         208      4200 female 2008
NA     <NA>   <NA>       NA       NA          NA        NA   <NA>   NA
```

We use the first position inside the square brackets to specify rows and the second position to specify columns.
In the first position, we have supplied this logical condition:

```r
penguins$species == "Gentoo" & penguins$body_mass < 4300
```

This compares each row with both criteria. The `&` operator means that both conditions must be `TRUE`.

Because the second position is empty, all columns are returned. If we do not want all the columns, we can specify the columns to keep in the second position:

```r
lowGentoo <- penguins[penguins$species == "Gentoo" & penguins$body_mass < 4300,
                      c("island", "flipper_len", "body_mass")]

print(lowGentoo)
```

```text
    island flipper_len body_mass
167 Biscoe         210      4200
169 Biscoe         210      4150
179 Biscoe         216      4100
193 Biscoe         208      3950
199 Biscoe         210      4200
213 Biscoe         208      4200
NA    <NA>          NA        NA
```

The rows are the same as before, but only the selected columns are returned.

If we want to select just the flipper lengths of penguins weighing less than 3,000 grams, we can write:

```r
penguins$flipper_len[penguins$body_mass < 3000]
```

```text
[1]  NA 179 187 181 184 178 193 188  NA 187 192
```

There is no comma in this expression because `penguins$flipper_len` is already a single vector. Vectors have one dimension, so no need for commas when indexing.

### `subset()`

R provides a function called `subset()` that can make subsetting data frames more convenient. The first argument is the data frame you want to subset. You then provide a condition specifying which rows to keep. The `select` argument can be used to specify which columns to keep.

We can use `subset()` to produce the same result as above:

```r
lowGentoo <- subset(
  penguins,
  species == "Gentoo" & body_mass < 4300,
  select = c("island", "flipper_len", "body_mass")
)

print(lowGentoo)
```

```text
    island flipper_len body_mass
167 Biscoe         210      4200
169 Biscoe         210      4150
179 Biscoe         216      4100
193 Biscoe         208      3950
199 Biscoe         210      4200
213 Biscoe         208      4200
```

This code is easier to read in part because we pass `penguins` directly to `subset()`. Within the filtering condition, R looks for the specified columns in that data frame, so we can write `species` and `body_mass` rather than `penguins$species` and `penguins$body_mass`.

The output contains the same observations and columns as the earlier example, but there is one important difference: `subset()` excludes rows for which the condition evaluates to `NA`. It keeps only rows where the condition is explicitly `TRUE`.

For example, if either `species` or `body_mass` is missing, the filtering condition evaluates to `NA`, and `subset()` excludes that row.
