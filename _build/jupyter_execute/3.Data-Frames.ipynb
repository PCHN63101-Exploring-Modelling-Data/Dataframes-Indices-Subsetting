## Data frames

So far, the objects we have looked at in R have been vectors and single values. Most of the time, you will work with data frames. A data frame is a two-dimensional data structure organised into rows and columns. Data from an experiment, for example, might be imported into R and stored in one or more data frames, with one row for each participant or trial and one column for each measure, predictor, or other variable.

As far as R is concerned, a data frame is essentially a collection of vectors. Each column usually contains values of a consistent type, such as numeric, character, logical, or factor. However, different columns can have different types. This allows you to keep reaction times, survey responses, ages, names, and experimental conditions together in a single data frame. All columns must have the same length, meaning that they must contain the same number of rows.

You may encounter the following terminology where data frames are concerned. Rows are often referred to as *observations*, while columns are referred to as *variables*. Cells, where observations and variables intersect contain *values*.

```{tip}
There are other two-dimensional objects in R and it's worth understanding the differences, although you will be working predominantly with data frames. You may also encounter:
* Table: In R, `table()` is a function that creates a contingency table. Its output is an object of class `"table"` containing counts of the combinations of values in one or more categorical vectors.
* Matrix: A matrix is a two-dimensional object whose elements must all have the same underlying data type, such as numeric, character, or logical. Matrices are commonly used for mathematical and linear-algebra operations
* Tibble: A tibble is a modern version of a data frame, provided by the `tibble` package and commonly used in the tidyverse. It preserves the basic rows-and-columns structure of a data frame but has some different behaviours, particularly when printing and subsetting. We will return to tibbles later in the unit.
```

### Creating a data frame

We will often read data into a data frame from the web or a csv file. But we can also create them in R. Let's make up some data from an experiment:

```r
results <- data.frame(
    participant = c('P1','P2','P3','P4','P5','P1','P2','P3','P4','P5'),
    reactionTime =  c(2.33,1.21,2.00,1.96,1.21,3.33,2.11,3.11,2.36,2.21),
    condition = c('con','con','con','con','con','inc','inc','inc','inc','inc'),
    correct = c(TRUE,TRUE,TRUE,TRUE,TRUE,FALSE,TRUE,TRUE,TRUE,TRUE)
)

print(results)
```


```{tip}
A couple of things are worth noting about the code above. I’ve split the `data.frame()` call across multiple lines to make it easier to read. Although it appears on several lines on screen, R treats it as a single expression. It knows to continue reading because the opening parenthesis `(` has not yet been closed by a corresponding closing parenthesis `)`.

We could have written the entire call on one long line. This would make no difference to R, but it would be much harder for us to read.

The indentation is also not necessary for the code to run, unlike in Python. It is mainly there to improve readability. RStudio automatically adds indentation when you write code in a script, but it does not usually format multiline expressions when you type directly into the console.
```

```text
   participant reactionTime condition correct
1           P1         2.33       con    TRUE
2           P2         1.21       con    TRUE
3           P3         2.00       con    TRUE
4           P4         1.96       con    TRUE
5           P5         1.21       con    TRUE
6           P1         3.33       inc   FALSE
7           P2         2.11       inc    TRUE
8           P3         3.11       inc    TRUE
9           P4         2.36       inc    TRUE
10          P5         2.21       inc    TRUE
```

In the code above we have created a new data frame called `results` and then printed it to the console. We can verify that it is indeed a data frame using the `class()` function:

```r
class(results)
```
```text
[1] "data.frame"
```
You could also combine several related vectors of the same length by passing the names of those vectors to `data.frame()`:
```r
minTemp <- c(2, 4, 3, 8)
month <- c('Jan', 'Feb', 'Mar', 'Apr')
rain <- c(15,14,20,10)

weather <- data.frame(month,minTemp,rain)
print(weather)
```

```text
  month minTemp rain
1   Jan       2   15
2   Feb       4   14
3   Mar       3   20
4   Apr       8   10
```

### Inspecting a data frame

R provides several functions that help us learn about the properties of a data frame. First, we can find out how many rows it contains using `nrow()`:

```r
nrow(results)
```

```text
[1] 10
```

We can find the number of columns using `ncol()`:

```r
ncol(results)
```

```text
[1] 4
```

Alternatively, `dim()` returns both dimensions at once. The result gives the number of rows first and the number of columns second:

```r
dim(results)
```

```text
[1] 10  4
```

We can also obtain the names of all the columns using `names()`:

```r
names(results)
```

```text
[1] "participant"  "reactionTime" "condition"    "correct"
```

Functions that tell us about the structure of a data frame are often particularly useful. For example, `str()` displays the object’s internal structure:

```r
str(results)
```

```text
'data.frame':	10 obs. of  4 variables:
 $ participant : chr  "P1" "P2" "P3" "P4" ...
 $ reactionTime: num  2.33 1.21 2 1.96 1.21 3.33 2.11 3.11 2.36 2.21
 $ condition   : chr  "con" "con" "con" "con" ...
 $ correct     : logi  TRUE TRUE TRUE TRUE TRUE FALSE ...
```

This output gives us several useful pieces of information:

- The object is a `data.frame`.
- It has 10 observations, or rows.
- It has 4 variables, or columns.
- It shows the name of each variable.
- It shows the data type of each variable:
  - `chr` means character.
  - `num` means numeric.
  - `logi` means logical.
- It displays some of the values in each variable. The `...` indicates that additional values have been omitted from the display.

Inspecting a data frame in this way is a useful first step after importing or creating data. It helps us check that the data have the expected dimensions, column names, data types, and values.
