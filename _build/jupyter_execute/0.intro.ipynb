# Introduction: Dataframes Indices & Subsetting

This week, we will develop a crucial skill for working in R and analysing data: extracting and filtering data. You will often work with large datasets, but a particular analysis may require only a small portion of the data. It is therefore important to be able to extract the relevant data efficiently and accurately.

We will begin with vectors, learning how to extract values based on their position, or *index*. We will then introduce data frames, the data structure you are likely to use most often when analysing data. Finally, we will cover how to extract specific columns from a data frame and filter rows according to particular criteria.
