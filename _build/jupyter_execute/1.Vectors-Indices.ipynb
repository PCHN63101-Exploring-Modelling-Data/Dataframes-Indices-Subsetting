# Vectors and Indices

As we saw last week, a vector is a collection of values of the same type (e.g, numeric, character, logical). Let's make a new vector with some pretend rainfall data for the last 12 months:

```r
rainfall <- c(12, 14, 10, 20, 8, 8, 6, 8, 12, 14, 14, 12)
```
We can check it's length and type using the `length()` and `class()` functions like so:

```r
length(rainfall)
```
which produces:
```text
[1] 12
```
```r
class(rainfall)
```
```text
[1] "numeric"
```
So we know the vector contains 12 numeric values, We are assuming that the first value represents January, the second February, and so on. Now suppose we want to know the rainfall for June. June will be the sixth element in the vector. It's easy to extract a value from a vector if you know it's position. You can do this using square brackets: `[]` 

```r
rainfall[6]
```
which generates the following output:
```text
[1] 8
```
The number in the square brackets told R the position, or index, of the element we wanted. 


### Extracting multiple values

You can extract more than one value at a time. To obtain the rainfall values for the first three months of the year, we need to give R the indices `1`, `2`, and `3` inside the square brackets.

You might be tempted to write:

```r
rainfall[1, 2, 3]
```

However, this will not work. Commas inside square brackets separate different dimensions. This expression attempts to index three dimensions, but a vector has only one dimension. As a result, R returns an error.

Instead, we need to provide the indices as a single vector:

```r
rainfall[c(1, 2, 3)]
```

```text
[1] 12 14 10
```

The `c()` function combines its arguments into a vector. Because `1`, `2`, and `3` are consecutive numbers, we can write this more compactly using the colon operator:

```r
rainfall[1:3]
```

```text
[1] 12 14 10
```

Here, `1:3` creates the index vector `c(1, 2, 3)`.

In R, there are often several ways to achieve the same result. For example, all of the following expressions return the first six values of `rainfall`:

```r
rainfall[c(1, 2, 3, 4, 5, 6)]
rainfall[1:6]
rainfall[seq(1, 6, 1)]
rainfall[c(TRUE, TRUE, TRUE, TRUE, TRUE, TRUE, FALSE, FALSE, FALSE, FALSE, FALSE, FALSE)]
```

Some approaches are more concise and readable than others. In general, choose the method that communicates your intention most clearly. Here, `rainfall[1:6]` is the simplest option.

The final example uses a logical index: it keeps the first six elements because their corresponding values are `TRUE` and excludes the remaining elements because their values are `FALSE`. Logical indexing becomes particularly useful when selecting values based on conditions rather than fixed positions.

### Negative indexing

Sometimes it is easier to say which values you don't want rather than which values you do want. Suppose we wanted rainfall values for all the months except January. We can use the minus sign `-` to achieve this. For example:

```r
rainfall[-1]
```
```text
[1] 14 10 20  8  8  6  8 12 14 14 12
```
`[-1]` tells R we want all the values except the first one. This would work just as well with a range, so if we wanted to exclude July to December we would use:

```r
rainfall[-(7:12)]
```
```text
[1] 12 14 10 20  8  8
```
The parentheses make it clear that we first create the sequence `7:12` and then make each index negative.
