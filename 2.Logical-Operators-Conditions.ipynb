# Logical Operators and Conditions

We often want to extract values according to a criterion or threshold rather than their position. Suppose that rainfall below 10 units is considered a drought. Were there any such months in our `rainfall` vector?

We can test this using the less-than operator, `<`:

```r
rainfall < 10
```
```text
 [1] FALSE FALSE FALSE FALSE  TRUE  TRUE  TRUE  TRUE FALSE FALSE FALSE FALSE
```
The result is a logical vector containing one `TRUE` or `FALSE` value for each element of `rainfall`. Each value indicates whether the corresponding rainfall value is less than 10.

There are four `TRUE` values, so four months had rainfall below 10.

Logical values can also be coerced to numbers in R: `TRUE` becomes `1` and `FALSE` becomes `0`. This means we can count the number of `TRUE` values using `sum()`:

```r
sum(rainfall < 10)
```
```text
[1] 4
```

```{tip}
When writing code with operators such as `+`, `-`, `/`, and `<`, place a space on either side of the operator. R does not require these spaces, but using them consistently makes code easier to read.

### Logical operators

R provides several logical operators and comparison operators:

| Operator | Meaning |
|---|---|
| `&` | AND |
| `\|` | OR |
| `!` | NOT |
| `==` | Equal to |
| `<` | Less than |
| `>` | Greater than |
| `<=` | Less than or equal to |
| `>=` | Greater than or equal to |

When you use `&` or `|` with vectors, R checks the conditions one position at a time. For each position, `&` returns `TRUE` only when both conditions are `TRUE`, while `|` returns `TRUE` when at least one condition is `TRUE`.

Suppose we want to identify rainfall values below 10 **or** above 13:

```r
rainfall < 10 | rainfall > 13

```
```text
 [1] FALSE  TRUE FALSE  TRUE  TRUE  TRUE  TRUE TRUE FALSE  TRUE  TRUE FALSE
```
The result is a logical vector with one value for each element of rainfall indicating whether that value is less than 10 or greater than 13.

Suppose we wanted to identify people of working age, let's say 18 or over and less than 67. Let's first create a vector of ages:
```r
ages <- c(16, 19, 15, 18, 18, 25, 32, 45, 68, 71, 67, 66)
print(ages)
```
```text
[1] 16 19 15 18 18 25 32 45 68 71 67 66
```
Note that I have used the `print()` function which simply prints whatever object you give it to the console. We could also have just typed `ages` into the console and got the same effect, but using `print()` hopefully makes it a little easier for you to follow the code. As ever, there's more than one way of doing things.

Now let's see which elements of the `ages` vector meet our criteria:

```r
ages >= 18 & ages < 67
```
```text
 [1] FALSE  TRUE FALSE  TRUE  TRUE  TRUE  TRUE TRUE FALSE FALSE FALSE  TRUE
```
The condition is evaluated element by element. For an element to produce `TRUE`, both comparisons must be `TRUE`.

You cannot write the condition as:

```r
ages >= 18 & < 67
```
Although this is understandable in English, it is incomplete in R. After `&`, R needs another complete comparison. It does not assume that `< 67` refers to `ages`; the object being compared must be stated explicitly.

````{warning}
When using equality comparisons, be careful to use `==`, not `=`.

The operator `==` tests whether two values are equal. The operator `=` is primarily used for assignment, just like `<-`:

```r
age <- 16
age == 18
```

```text
[1] FALSE
```

This works as intended. We first assign the value `16` to `age`, then ask whether `age` is equal to `18`. R returns `FALSE`.

If we use `=` instead of `==`:

```r
age <- 16
age = 18
```

R assigns the value `18` to `age`. It does not compare the two values. Assignments do not normally produce visible output in the console, but we can check the value of `age`:

```r
print(age)
```

```text
[1] 18
```

Therefore:

- `==` means “is equal to?”
- `=` means “assign this value” in this context
- `<-` also means “assign this value”

The `=` operator is also commonly used to provide named arguments in function calls:

```r
round(3.14159, digits = 2)
```

Here, `digits = 2` does not assign a new object called `digits`; it supplies `2` as the `digits` argument to `round()`.

````



### Logic as a filter

Logical tests are particularly useful for extracting data. Suppose we want to create a new vector containing only ages greater than 67. As we saw above, we can compare each value in `ages` with 67:

```r
ages > 67
```

```text
[1] FALSE FALSE FALSE FALSE FALSE FALSE FALSE FALSE  TRUE  TRUE FALSE FALSE 
```

The result is a logical vector containing one `TRUE` or `FALSE` value for each element of `ages`. We can use this logical vector as a **filter** or **mask** to select values:

```r
retired <- ages[ages > 67]
print(retired)
```

```text
[1] 68 71
```

Here, `ages > 67` produces the logical vector used inside the square brackets. R keeps the values of `ages` corresponding to `TRUE` and excludes the values corresponding to `FALSE`. The selected values are then assigned to the new vector, `retired`.

In other words:

```r
ages[ages > 67]
```

means “return the elements of `ages` for which `ages > 67` is `TRUE`.”

### Missing values

Dealing with missing values is a recurring theme in R and data analysis more generally. Most real-world datasets contain missing values, and R requires us to handle them explicitly.

What happens when we perform a logical comparison on a vector that contains missing values? Here is some hypothetical temperature data. Suppose we want to identify values below freezing:

```r
temperature <- c(-10, -8, -4, 0, 2, NA, 2, 4, 8, 12, NA)

temperature < 0
```

```text
 [1]  TRUE  TRUE  TRUE FALSE FALSE    NA FALSE FALSE FALSE FALSE    NA
```

The comparison returns `NA` for the missing values because R cannot determine whether an unknown temperature is below zero.

What happens when we use this result as a logical filter?

```r
freezing <- temperature[temperature < 0]

print(freezing)
```

```text
[1] -10  -8  -4  NA  NA
```

The missing values are included in the new `freezing` vector. This happens because the logical filter contains `NA` values, and R cannot determine whether the corresponding elements should be included or excluded.

The same issue would occur if we created a vector of temperatures at or above freezing:

```r
aboveFreezing <- temperature[temperature >= 0]

print(aboveFreezing)
```

```text
[1]  0  2 NA  2  4  8 12 NA
```

The missing values are included here too. This is usually not what we want, so we should identify missing values and deal with them explicitly.

A useful function for this purpose is `is.na()`. It checks whether each element of an object is `NA`:

```r
is.na(temperature)
```

```text
 [1] FALSE FALSE FALSE FALSE FALSE  TRUE FALSE FALSE FALSE FALSE  TRUE
```

The result is a logical vector containing `TRUE` for each missing value and `FALSE` for each non-missing value.

We can now create a filter that selects values that are both not missing and below zero:

```r
freezing <- temperature[!is.na(temperature) & temperature < 0]

print(freezing)
```

```text
[1] -10  -8  -4
```

Let’s unpack this condition:

```r
!is.na(temperature) & temperature < 0
```

In English, this means:

> Select values that are not missing and are below zero.

The first part,

```r
!is.na(temperature)
```

identifies values that are not `NA`. The function `is.na()` returns `TRUE` for missing values, so the `!` operator reverses those results: `TRUE` becomes `FALSE`, and `FALSE` becomes `TRUE`.

The `&` operator means that both conditions must be `TRUE`:

1. The value is not missing.
2. The value is below zero.

We then place this logical condition inside square brackets to tell R which elements of `temperature` to extract.

Don't worry if this isn't immediately intuitive, R is primarily learned through writing code and fixing errors. In other words, it takes practice.
